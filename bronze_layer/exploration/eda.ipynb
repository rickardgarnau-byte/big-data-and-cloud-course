```python
from pyspark.sql import functions as F

df.select([
    F.sum(F.col(c).isNull().cast("int")).alias(c)
    for c in df.columns
]).show()
```

```python
df.select(F.sum(F.col("Customer Password").isNull().cast("int"))).show()
```

```python
from pyspark.sql import functions as F

total = df.count()
unique = df.dropDuplicates().count()

print(f"Totalt: {total}")
print(f"Unika rader: {unique}")
print(f"Dubbletter: {total - unique}")
```
